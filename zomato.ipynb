```python
cuisines_f=data.groupby('Cuisines')['Average Cost for two'].mean().sort_values(ascending=False).reset_index()
```

```python
data.merge(cuisines_f,left_on='Cuisines',right_on='Cuisines')
```

```python
data=data.merge(cuisines_f,left_on='Cuisines',right_on='Cuisines')
```

```python
data.drop(columns=['Cuisines'],inplace=True)
```

```python
data.head(50)
```

```python
data.rename(columns={'Average Cost for two_x':'cost','Average Cost for two_y':'cuisine'},inplace=True)
```

```python
y=data.iloc[:,1].values
y.shape
```

```python
y
```

```python
x=data.drop(columns=['cost']).iloc[:,:].values
```

```python
x
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
x[:,0]=encoder.fit_transform(x[:,0])
```

```python
x[:,0]
```

```python
x.shape
```

```python
x[:,0:4]
```

```python
x=x[:,1:]
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
regressor1=LinearRegression()
```

```python
regressor1.fit(x_train,y_train)
```

```python
from sklearn.metrics import r2_score
```

```python
y_pred=regressor1.predict(x_test)
```

```python
r2_score(y_test,y_pred)*100
```

```python
y_test
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
print(f"Mean Squared Error: {mse}")
```
